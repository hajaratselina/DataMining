```python
import pandas as pd

df_clusters = pd.read_csv('kmeans_market_distribution.csv')
df_raw = pd.read_csv('../01 - DataPreparation/board_games.csv')
```

```python
df_merged = df_raw.merge(df_clusters, on='BGGId')
df_merged
```

```python
for cluster in range(0,9):
    current_cluster = df_merged.loc[df_merged['cluster']==cluster]
    print(current_cluster.loc[current_cluster['NumOwned']<=5000].sort_values(by='NumOwned', ascending=False).head(50)['Name'])
    # print(current_cluster.sort_values(by='NumOwned', ascending=True).head(100)['Name'])
    print("="*80)
```
